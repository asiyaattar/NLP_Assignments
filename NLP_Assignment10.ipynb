```python
import nltk
from nltk.util import ngrams
from collections import defaultdict, Counter
from nltk.tokenize import word_tokenize
import random

# Download NLTK tokenizer
nltk.download('punkt_tab') # Changed to download 'punkt_tab' as suggested by the error

# Sample corpus (replace this with a larger corpus for better results)
corpus = """
Natural language processing enables computers to understand human language.
Language models are used in translation, text prediction, and speech recognition.
N-gram models predict the next word based on the previous words.
Auto-complete algorithms make typing faster and smarter.
"""

# Tokenize the corpus
tokens = word_tokenize(corpus.lower())

# Build N-gram models (bigram and trigram)
def build_ngram_model(tokens, n=2):
    model = defaultdict(Counter)
    for ngram in ngrams(tokens, n, pad_left=True, pad_right=True):
        prefix = ngram[:-1]  # all words except last
        next_word = ngram[-1] # the word to predict
        model[prefix][next_word] += 1
    return model

# Bigram (2-gram) model
bigram_model = build_ngram_model(tokens, n=2)
# Trigram (3-gram) model
trigram_model = build_ngram_model(tokens, n=3)

# Function to predict next word using N-gram model
def predict_next_word(context, model, n=2):
    context = tuple(word_tokenize(context.lower())[-(n-1):])
    if context in model:
        # Return the most common next word
        return model[context].most_common(1)[0][0]
    else:
        return random.choice(list(model[random.choice(list(model.keys()))].keys()))

# Test auto-complete
test_sentences = [
    "language models",
    "auto-complete",
    "n-gram models predict"
]

print("=== Auto-complete using N-gram model ===")
for sentence in test_sentences:
    next_word_bigram = predict_next_word(sentence, bigram_model, n=2)
    next_word_trigram = predict_next_word(sentence, trigram_model, n=3)
    print(f"Input: '{sentence}'")
    print(f"Bigram Prediction: {next_word_bigram}")
    print(f"Trigram Prediction: {next_word_trigram}")
    print("-" * 50)
```
