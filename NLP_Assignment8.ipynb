```python
# Install necessary library if not already installed
# !pip install nltk

import nltk
from nltk.corpus import wordnet as wn
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

# Download required NLTK data
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt')

# Example ambiguous sentence with clear context
sentence = "John went to the bank to deposit his paycheck at the financial institution."

# Tokenize the sentence
tokens = word_tokenize(sentence)

# Apply Lesk algorithm to disambiguate the word 'bank'
sense = lesk(tokens, 'bank', 'n')  # 'n' for noun

# Output the result
if sense:
    print("Word:", 'bank')
    print("Disambiguated Sense:", sense)
    print("Definition:", sense.definition())
    print("Example:", sense.examples())
else:
    print("No suitable sense found.")
```
